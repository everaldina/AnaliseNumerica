# Exercicio 

Calcular a interpolação cubica de splines para a função (f(x) = cos(x)) no intervalo \([0, 1]\), considerando as condições de contorno natural e fixada. Comparar os resultados obtidos com os valores reais da função em pontos intermediários do intervalo.

```python
from splines import get_splines
from utils import obter_indice_intervalo
import sympy as sp
```

```python
x_a = [0, 0.25, 0.5, 0.75, 1]
x_b = [0, 0.125, 0.25, 0.375, 0.5, 0.625, 0.75, 0.875, 1]
expression = 'cos(pi * x)'
x_eval = 0.5
interval = (0, 1)
```

```python
# integral veradeira intervalo [a, b]
expression_sym = sp.sympify(expression)
integral = sp.integrate(expression_sym, (sp.symbols('x'), interval[0], interval[1]))
print(f'Integral verdadeira: {integral.evalf()}')
```

```python
# derivada verdadeira no ponto x_eval
expression_deriv = sp.diff(expression_sym, sp.symbols('x'))
deriv_true = expression_deriv.evalf(subs={sp.symbols('x'): x_eval})
print(f'Derivada verdadeira em x={x_eval}: {deriv_true}')
```

```python
# segunda derivada verdadeira
expression_2deriv = sp.diff(expression_deriv, sp.symbols('x'))
deriv2_true = expression_2deriv.evalf(subs={sp.symbols('x'): x_eval})
print(f'Segunda derivada verdadeira em x={x_eval}: {deriv2_true}')
```

## Spline natural

### Expressoes

```python
splines_a = get_splines(expression, x_a, condicao='natural')
splines_b = get_splines(expression, x_b, condicao='natural')
```

```python
print("Splines com condição de contorno natural (x_a):")
for i, spline in enumerate(splines_a):
    print(f"Spline {i}: {spline}")
```

```python
print("Splines com condição de contorno natural (x_b):")
for i, spline in enumerate(splines_b):
    print(f"Spline {i}: {spline}")
```

### Integrais

```python
# integral aproximada usando os splines
integral_spline = 0
for i, spline in enumerate(splines_a):
    a = x_a[i]
    b = x_a[i+1]
    integral_spline += sp.integrate(spline, (sp.symbols('x'), a, b))
    print(f'Integral aproximada do spline {i} no intervalo [{x_a[0]}, {b}]: {integral_spline.evalf()}')
    
print(f'Integral aproximada total usando splines (x_a): {integral_spline}')
```

```python
# integral aproximada usando os splines
integral_spline = 0
for i, spline in enumerate(splines_b):
    a = x_b[i]
    b = x_b[i+1]
    integral_spline += sp.integrate(spline, (sp.symbols('x'), a, b))
    print(f'Integral aproximada do spline {i} no intervalo [{x_a[0]}, {b}]: {integral_spline.evalf()}')
    
print(f'Integral aproximada total usando splines (x_b): {integral_spline}')
```

### Derivadas

```python
# avaliacao aproximada da derivada no ponto x_eval usando os splines
index_a = obter_indice_intervalo(x_a, x_eval)
spline_a = splines_a[index_a]
deriv_approx_a = sp.diff(spline_a, sp.symbols('x')).evalf(subs={sp.symbols('x'): x_eval})
print(f'Derivada aproximada em x={x_eval} usando splines (x_a): {deriv_approx_a}')
```

```python
## segunda derivada
derivada2_approx_a = sp.diff(spline_a, sp.symbols('x'), 2).evalf(subs={sp.symbols('x'): x_eval})
print(f'Segunda derivada aproximada em x={x_eval} usando splines (x_a): {derivada2_approx_a}')
```

```python
# avaliacao aproximada da derivada no ponto x_eval usando os splines
index_b = obter_indice_intervalo(x_b, x_eval)
spline_b = splines_b[index_b]
deriv_approx_b = sp.diff(spline_b, sp.symbols('x')).evalf(subs={sp.symbols('x'): x_eval})
print(f'Derivada aproximada em x={x_eval} usando splines (x_b): {deriv_approx_b}')
```

```python
# segunda derivada
derivada2_approx_b = sp.diff(spline_b, sp.symbols('x'), 2).evalf(subs={sp.symbols('x'): x_eval})
print(f'Segunda derivada aproximada em x={x_eval} usando splines (x_b): {derivada2_approx_b}')
```
