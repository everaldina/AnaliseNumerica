## Bissecção

```python
def print_results(result, dict_entrada):
    print("MÉTODO DA BISSECÇÃO - RESULTADOS")
    print("="*100)
    print(f"Função: {dict_entrada['expressao']}")
    print(f"Intervalo: [{dict_entrada['a']}, {dict_entrada['b']}], Precisão: {dict_entrada['precisao']}, Iterações Máximas: {dict_entrada['iteracoes']}")
    print("-"*100)
    
    # Cabeçalho
    print(f"{'k':>3} | {'a':>10} | {'b':>10} | {'f(a)':>12} | {'f(b)':>12} | {'xk':>12} | {'f(xk)':>12} | {'Erro Rel':>10}")
    print("-"*100)
    
    # Uma linha por iteração
    for k, v in result.items():
        print(f"{k:>3} | {float(v['a']):>10.6f} | {float(v['b']):>10.6f} | {float(v['f(a)']):>12.6f} | {float(v['f(b)']):>12.6f} | {float(v['xk']):>12.6f} | {float(v['f(xk)']):>12.6f} | {float(v['desvio_relativo']):>10.2e}")
    
    print("-"*100)
    ultima = result[max(result.keys())]
    print(f"Raiz aproximada: x = {float(ultima['xk'])}, f(x) = {float(ultima['f(xk)']):.2e}, Iterações: {max(result.keys())}")
```

```python
import csv

def save_results(result, dict_entrada, path):
    # Cabeçalho do CSV
    header = ['k', 'a', 'b', 'f(a)', 'f(b)', 'xk', 'f(xk)', 'Erro Rel']
    with open(path, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        # Informações iniciais como comentários
        writer.writerow([f"# Função: {dict_entrada['expressao']}"])
        writer.writerow([f"# Intervalo: [{dict_entrada['a']}, {dict_entrada['b']}], Precisão: {dict_entrada['precisao']}, Iterações Máximas: {dict_entrada['iteracoes']}"])
        writer.writerow([])  # linha em branco
        writer.writerow(header)
        for k, v in result.items():
            writer.writerow([
                k,
                float(v['a']),
                float(v['b']),
                float(v['f(a)']),
                float(v['f(b)']),
                float(v['xk']),
                float(v['f(xk)']),
                float(v['desvio_relativo'])
            ])
```

```python
epsilon = "1e-10" # soma ou subtração de 1e-15 pois os valores de f(a) e f(b) vão dar uma divisão por zero
pasta_saida = "lista_2/bisseccao"
if not os.path.exists(pasta_saida):
    os.makedirs(pasta_saida)
```

### Para a = 1

```python
conjuntos_intervalos = [
    (f'pi/2 + {epsilon}', f'3*pi/2 - {epsilon}'),
    (f'3*pi/2 + {epsilon}', f'5*pi/2 - {epsilon}'),
    (f'5*pi/2 + {epsilon}', f'7*pi/2 - {epsilon}'),
    (f'7*pi/2 + {epsilon}', f'9*pi/2 - {epsilon}'),
]

for i, intervalo in enumerate(conjuntos_intervalos):
    a, b = intervalo
    entrada_dict = {
        'a': a,
        'b': b,
        'precisao': 10**-10,
        'expressao': 'x/1 - tan(1*x)',
        'iteracoes': 100
    }
    resultados = run_bisseccao(entrada_dict=entrada_dict)
    if resultados is not None:
        print_results(resultados, entrada_dict)
        save_results(resultados, entrada_dict, path=os.path.join(pasta_saida, f"a1_I{i}.csv"))
        print("\n\n")
    else:
        print(f"Não foi possível encontrar a raiz no intervalo [{a}, {b}]")



```
