```python
def print_results(result, dict_entrada, metodo="Bissecção"):
    print(f"MÉTODO DA {metodo.upper()} - RESULTADOS")
    print("="*100)
    print(f"Função: {dict_entrada['expressao']}")
    print(f"Intervalo: [{dict_entrada['a']}, {dict_entrada['b']}], Precisão: {dict_entrada['precisao']}, Iterações Máximas: {dict_entrada['iteracoes']}")
    print("-"*100)
    
    # Cabeçalho
    print(f"{'k':>3} | {'a':>10} | {'b':>10} | {'f(a)':>12} | {'f(b)':>12} | {'xk':>12} | {'f(xk)':>12} | {'Erro Rel':>10}")
    print("-"*100)
    
    # Uma linha por iteração
    for k, v in result.items():
        print(f"{k:>3} | {float(v['a']):>10.6f} | {float(v['b']):>10.6f} | {float(v['f(a)']):>12.6f} | {float(v['f(b)']):>12.6f} | {float(v['xk']):>12.6f} | {float(v['f(xk)']):>12.6f} | {float(v['desvio_relativo']):>10.2e}")
    
    print("-"*100)
    ultima = result[max(result.keys())]
    print(f"Raiz aproximada: x = {float(ultima['xk'])}, f(x) = {float(ultima['f(xk)']):.2e}, Iterações: {max(result.keys())}")
```

```python
import csv

def save_results(result, dict_entrada, path):
    # Cabeçalho do CSV
    header = ['k', 'a', 'b', 'f(a)', 'f(b)', 'xk', 'f(xk)', 'Erro Rel']
    with open(path, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        # Informações iniciais como comentários
        writer.writerow([f"# Função: {dict_entrada['expressao']}"])
        writer.writerow([f"# Intervalo: [{dict_entrada['a']}, {dict_entrada['b']}], Precisão: {dict_entrada['precisao']}, Iterações Máximas: {dict_entrada['iteracoes']}"])
        writer.writerow([])  # linha em branco
        writer.writerow(header)
        for k, v in result.items():
            writer.writerow([
                k,
                float(v['a']),
                float(v['b']),
                float(v['f(a)']),
                float(v['f(b)']),
                float(v['xk']),
                float(v['f(xk)']),
                float(v['desvio_relativo'])
            ])
```

# Metodos abertos

```python
def save_results_abertos(result, dict_entrada, path):
    # Cabeçalho do CSV
    header = ['k', 'xk', 'f(xk)', 'Erro Rel']
    with open(path, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        # Informações iniciais como comentários
        writer.writerow([f"# Função: {dict_entrada['expressao']}"])
        if 'x1' not in dict_entrada:
            writer.writerow([f"# Chute inicial: {dict_entrada['x0']}, Precisão: {dict_entrada['precisao']}, Iterações Máximas: {dict_entrada['iteracoes']}"])
        else:
            writer.writerow([f"# Chutes iniciais: {dict_entrada['x0']}, {dict_entrada['x1']}, Precisão: {dict_entrada['precisao']}, Iterações Máximas: {dict_entrada['iteracoes']}"])
        writer.writerow([])  # linha em branco
        writer.writerow(header)
        for k, v in result.items():
            writer.writerow([
                k,
                float(v['xk']),
                float(v['f(xk)']),
                float(v['desvio_relativo'])
            ])
```

## Ponto fixo


- f(x) = x/a+tan(ax) = 0
    => x = -atan(ax)
    => g(x) = -atan(ax)
    => g'(x) = -a^2*sec^2(ax)
- Para a = 1
    - g(x) = -tan(x)
    - g'(x) = -sec^2(x)
- Para a = 2
    - g(x) = -2tan(2x)
    - g'(x) = -2*sec^2(2x)


```python
chutes = {
    'a1': [],
    'a2': []
}
for i in range(len(conjunto_intervalos['a1'])):
    a1, b1 = conjunto_intervalos['a1'][i]
    a2, b2 = conjunto_intervalos['a2'][i]
    chutes['a1'].append( (f'({a1} + {b1})/2'))
    chutes['a2'].append( (f'({a2} + {b2})/2'))
```

```python
pasta_saida = os.path.join(pasta_root, 'ponto_fixo')
if not os.path.exists(pasta_saida):
    os.makedirs(pasta_saida)
```

### Para a = 1

```python
for i, chute in enumerate(chutes['a1']):
    a, b = intervalo
    entrada_dict = {
        'x0': chute,
        'precisao': precisao,
        'expressao': '-tan(x)',
        'iteracoes': iteracoes
    }
    resultados = run_ponto_fixo(entrada_dict=entrada_dict)
    ultima = resultados[max(resultados.keys())]
    if abs(ultima['f(xk)']) < precisao: 
        print_results_abertos(resultados, entrada_dict, metodo="Ponto Fixo")
        save_results_abertos(resultados, entrada_dict, path=os.path.join(pasta_saida, f"a1_C{i}.csv"))
        print("\n\n")
    else:
        print(f"Não foi possível encontrar a raiz no intervalo [{a}, {b}]")
```

```python
for i, chute in enumerate(chutes['a1']):
    a, b = intervalo
    entrada_dict = {
        'x0': chute,
        'precisao': precisao,
        'expressao': '-2*tan(2*x)',
        'iteracoes': iteracoes
    }
    resultados = run_ponto_fixo(entrada_dict=entrada_dict)
    ultima = resultados[max(resultados.keys())]
    if abs(ultima['f(xk)']) < precisao: 
        print_results_abertos(resultados, entrada_dict, metodo="Ponto Fixo")
        save_results_abertos(resultados, entrada_dict, path=os.path.join(pasta_saida, f"a2_C{i}.csv"))
        print("\n\n")
    else:
        print(f"Não foi possível encontrar a raiz no intervalo [{a}, {b}]")
```
